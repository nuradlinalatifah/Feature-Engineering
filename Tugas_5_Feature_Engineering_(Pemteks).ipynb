```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

```python
# Membaca data hasil preprocessing
df = pd.read_csv('berita_kompas_prepro.csv')

df = df.dropna().reset_index().drop(columns=['index'])

df_pre = df['Prepro']

df_pre.info()
```

```python
# Download tokenizer NLTK
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Memecah teks menjadi token
from nltk import word_tokenize
from tqdm import tqdm

sentences = [word_tokenize(tweet.lower()) for tweet in tqdm(df_pre)]

sentences
```

#Build Word2Vec Model

```python
# Membangun model Word2Vec
!pip install gensim

import gensim.models

model = gensim.models.Word2Vec(
    sentences,
    min_count=1,
    vector_size=100,
    window=5,
    sg=0
)
```

```python
# Menyimpan model Word2Vec
model.save('berita_kompas.w2v')
```

```python
# Memuat model Word2Vec
model = gensim.models.Word2Vec.load('berita_kompas.w2v')
```

```python
# Melihat kosakata dalam model
w2v = model.wv
w2v.index_to_key[:10]
```

```python
# Melihat jumlah kata dan dimensi vektor
w2v.vectors.shape, w2v.vector_size
```

```python
# Melihat kata-kata yang tersedia
print(w2v.index_to_key[:20])
```

```python
# Melihat kata-kata yang mirip dengan kata pertama
w2v.similar_by_key(w2v.index_to_key[0])
```

```python
# Mengubah dokumen menjadi vektor numerik
import numpy as np

def get_document_vector(tokens, model):
    valid_vectors = [
        model.wv[word]
        for word in tokens
        if word in model.wv
    ]
    return np.mean(valid_vectors, axis=0)

X_features = np.array([
    get_document_vector(doc, model)
    for doc in sentences
])

pd.DataFrame(X_features)
```

```python
# Visualisasi Word2Vec
!pip install umap-learn
```

```python
# Mengimpor library visualisasi
from umap import UMAP
import plotly.express as px
```

```python
# Mengubah vektor kata menjadi dua dimensi
X = UMAP().fit_transform(w2v.vectors)

df2 = pd.DataFrame(X, columns=['umap1', 'umap2'])

df2['word'] = w2v.index_to_key

fig = px.scatter(
    df2,
    x='umap1',
    y='umap2',
    text='word'
)

fig.update_traces(textposition='top center')

fig.update_layout(
    height=800,
    title_text='Word2Vec Visualization'
)

fig.show()
```

#Build FastText Model

```python
# Membangun model FastText
import gensim.models

model_fasttext = gensim.models.FastText(
    sentences,
    min_count=1,
    vector_size=100,
    window=5,
    min_n=3,
    max_n=6,
    sg=1,
    epochs=10
)
```

```python
# Menyimpan model FastText
model_fasttext.save('berita_kompas.fasttext')
```

```python
# Memuat model FastText
model_fasttext = gensim.models.FastText.load(
    'berita_kompas.fasttext'
)
```

```python
# Melihat kosakata dalam model
fasttext = model_fasttext.wv

fasttext.index_to_key[:10]
```

```python
 # Melihat jumlah kata dan dimensi vektor
fasttext.vectors.shape
```

```python
# Melihat ukuran dimensi vektor
fasttext.vector_size
```

```python
# Melihat kata-kata yang mirip
fasttext.similar_by_key(fasttext.index_to_key[0])
```

```python
# Mengubah dokumen menjadi vektor numerik
import numpy as np

def get_document_vector(tokens, model):
    valid_vectors = [
        model.wv[word]
        for word in tokens
        if word in model.wv
    ]
    return np.mean(valid_vectors, axis=0)

X_features_fasttext = np.array([
    get_document_vector(doc, model_fasttext)
    for doc in sentences
])

pd.DataFrame(X_features_fasttext)
```

```python
# Mengubah vektor FastText menjadi dua dimensi
X = UMAP().fit_transform(fasttext.vectors)

df2 = pd.DataFrame(X, columns=['umap1', 'umap2'])
df2['word'] = fasttext.index_to_key
```

```python
# Menampilkan visualisasi FastText
fig = px.scatter(
    df2,
    x='umap1',
    y='umap2',
    text='word'
)

fig.update_traces(textposition='top center')

fig.update_layout(
    height=800,
    title_text='FastText Visualization'
)

fig.show()
```

## Analisis Hasil Feature Engineering

Tokenization digunakan untuk memecah teks berita menjadi kata-kata sebagai input model.

Word2Vec mengubah kata menjadi vektor numerik berdimensi 100 berdasarkan konteksnya. Model menghasilkan 640 kata dalam vocabulary. Vektor kata kemudian dirata-ratakan untuk menghasilkan representasi numerik setiap dokumen.

UMAP digunakan untuk memvisualisasikan vektor kata dalam dua dimensi sehingga pola kedekatan antarkata dapat diamati.

FastText juga menghasilkan vektor berdimensi 100, tetapi turut mempertimbangkan potongan karakter kata (*character n-grams*). Vektor kata dirata-ratakan menjadi vektor dokumen dan divisualisasikan menggunakan UMAP.

**Kesimpulan:** Word2Vec dan FastText mengubah teks berita menjadi fitur numerik yang dapat digunakan untuk analisis teks dan pemodelan *machine learning*. Perbedaannya, FastText juga memanfaatkan informasi karakter kata.
